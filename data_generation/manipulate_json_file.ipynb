```python
def convert_html_to_json(html_content, index):
    # Parse HTML content
    soup = BeautifulSoup(html_content, 'html.parser')

    # Find the table
    table = soup.find('table')

    # Initialize an empty list to store table data
    table_data = []

    headers = True
    # Extract table data
    if index in indexes:
        headers = False
    rows = table.find_all('tr')
    for row_index, row in enumerate(rows):
        cols = row.find_all(['td'])[1:]
        if headers:
            if row_index == 0:
                headers = [col.text.strip() for col in cols]
            else:
                row_data = {}
                for i, (col, head) in enumerate(zip(cols, headers)):
                    if i == 0:
                        present_header = col.text.strip()
                        row_data[present_header] = {}
                    else:
                        row_data[present_header][head] = col.text.strip()
                table_data.append(row_data)
        else:
            row_data = {}
            for i, col in enumerate(cols):
                if i == 0:
                    present_header = col.text.strip()
                    row_data[present_header] = {}
                else:
                    row_data[present_header] = col.text.strip()
            table_data.append(row_data)

    # Convert list to JSON
    json_data = json.dumps(table_data, indent=4)

    # Print JSON data
    print(json_data)
```

```python
json_data = convert_html_to_json(odf.loc[1, 'html_content'], odf.loc[1, "index"])
```
